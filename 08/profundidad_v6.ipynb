**Stereo Depth Estimation**

```python
import cv2
import numpy as np
from matplotlib import pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
f= 5299.313
width=2988
height=2008
doffs=174.186
baseline=177.288 #in mm
ndisp=180
img1 = cv2.imread('im0.png',0) #queryimage # left image
img2 = cv2.imread('im1.png',0) #trainimage # right image
print(img1.shape[0])
plt.subplot(121),plt.imshow(img1)
plt.subplot(122),plt.imshow(img2)
plt.show()
```

```python
stereo = cv2.StereoBM_create(numDisparities=144, blockSize=11)
print(np.max(stereo))
disparity = stereo.compute(img1,img2)



plt.imshow(disparity, cmap='jet')
plt.show()

```

**Monocular Depth Estimation**

```python
from transformers import pipeline
import matplotlib.pyplot as plt
from PIL import Image
# load pipe
pipe = pipeline(task="depth-estimation", model="depth-anything/Depth-Anything-V2-Small-hf")
# load image
url = 'im0.png'
image = Image.open(url)
# inference
depth = np.array(pipe(image)["depth"])

plt.imshow(depth, cmap=plt.cm.jet)  
plt.title("Predicted Depth Map")
plt.show()
```
